Construcctor to improve counter code

```python
function Counter(){
    var count = 0; // enclosing in counter() to restrict its use
    this.incCount = function (){
        count++;
        console.log(count)
    }
    this.decCount = function(){
        count--;
        console.log(count)
    }
}

var c1 = new Counter();
c1.incCount();
c1.incCount();
c1.decCount();
```
