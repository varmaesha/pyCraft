```python
def rgb(r,g,b):
    return f"#{((r<<16)+(g<<8)+b):06X}"
print(rgb(0,51,225))
```

```python
def tran(mat):
    return[list(row) for row in zip(*mat)]
mat=[[1,2,3],[4,5,6],[7,8,9]]
print("\n".join(",".join(map(str,row)) for row in tran(mat)))
```

```python
from datetime import datetime
def is_valid_date(val):
    return datetime.strptime(val, "%B %d, %Y %H:%M:%S") if val else False
print(is_valid_date("December 14, 1995 03:24:12"))
```

```python
# DAY OF YEAR
from datetime import datetime
def day_of_year(date):
    return date.timetuple().tm_yday
print(day_of_year(datetime(1997,7,1)))
```

```python
from datetime import datetime
def day_diff(date1, date2):
    return abs((date2 - date1).days)+1
date1 = datetime(2020,10,21)
date2 = datetime(2020,1,9)
print(date1 -  date2)
```

```python
def cap(s):
    return s[0].upper() + s[1:] if s else ""
print(cap("follow for more"))
```

```python
# Character frequency
def char_feq(s):
    return {char: s.count(char) for char in set(s)}
print(",".join([f"{k}=>{v}" for k,v in char_feq("hello").items()]))
```

```python
import random
def ran_hx():
    return f"#{random.randint(0,0xFFFFFF):06X}"
print(ran_hx())
```

```python
import random
import string
def ran_str(lent):
    return ''.join(random.choice(string.ascii_lowercase) for _ in range(lent))
print(ran_str(7))
```
