```python
# Problem 1 : Take a variable x and print "Even" if the number is divisible by 2, otherwise print "Odd".
x=4

if(x%2==0):
    print("Even")
else:
    print("Odd")
```

```python
# Problem 2 : Take a variable y and print 
# "Grade A" if y is greater than 90, 
# "Grade B" if y is greater than 60 but less than or equal to 90 and 
# "Grade F" Otherwise.
y=90

if(y>90):
    print("Grade A")
elif(y>60):
    print("Grade B")
else:
    print("Grade F")
```

```python
# Problem : Create a function that takes two numbers as argument and returns the greater of the two.
def compare(a,b):
    if(a>b):
        greater=a
    else:
        greater=b
    
    return greater
print(compare(10,50))
print(compare(14,5))
             
```
