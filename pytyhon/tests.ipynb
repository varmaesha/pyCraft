```python
inventory = {'fruit': ['apple', 'banana', 'cherry'], 'price': (1.2, 1.3, 1.4)}

# Sorting the list of fruits
inventory['fruit'].sort()
print(inventory['fruit'], end=' ')

# Modifying the tuple of prices
P = list(inventory['price'])  # Convert tuple to list for modification
P[1] = 1.5                    # Change the second price
inventory['price'] = tuple(P)  # Convert list back to tuple
print(inventory['price'])
```

```python
import collections

nums = collections.defaultdict(lambda: 10)  # Create a defaultdict with a default value of 10

nums[1]  # Accessing key 1, which doesn't exist yet; it will be created with the default value (10)
nums[2] = 5  # Assign key 2 a value of 5

print(nums[3]) 
```

```python
def is_palindrome(s): 
    return s == s[::-1]  # Check if the string is equal to its reverse

palindrome_check = is_palindrome('radar')
print(f"Is 'radar' a palindrome? {palindrome_check}")
```

```python
class StringReverser:
    def __init__(self, text): 
        self.text = text  # Initialize the 'text' attribute with the given string

    def reverse(self): 
        self.text = self.text[::-1]  # Reverse the string and update the 'text' attribute

# Create an instance of StringReverser
sr = StringReverser('abc')

# Use getattr to get the 'reverse' method and call it
getattr(sr, 'reverse')()

# Print the reversed string
print(sr.text)
```

```python
def counter(n):
    if n == 0:  # Base case: when n reaches 0, stop recursion
        return 0
    else:
        return 2 + counter(n - 1)  # Recursive call reducing n by 1

n = 3
print(counter(n))
```

```python
class Profile:
    def __init__(self, name='Guest', active=False):
        self.name = name
        self.active = active  # Fix the assignment to self.active

    def status(self):
        print(self.name, 'Active:', self.active)  # Fix 'active' typo

# Create an instance of Profile and check its status
profile = Profile('John', True)
profile.status()  # Output will show the status of the profile
```

```python
import pandas as pd

# Creating the DataFrame
df = pd.DataFrame({
    'Fruit': ['Apple', 'Banana', 'Cherry'],
    'Quantity': [15, 10, 20],
    'Price': [1.5, 0.5, 2]
})

# Assigning the 'Total' column by multiplying 'Quantity' and 'Price'
df = df.assign(Total=df['Quantity'] * df['Price'])

# Displaying the DataFrame to see the result
print(df)

# Accessing the total value for Cherries
cherry_total = df.loc[df['Fruit'] == 'Cherry', 'Total'].values[0]
print(f"Total value for Cherries: {cherry_total}")
```

```python
class Modifier:
    def __init__(self):
        self.value = 'Start'
        self.change()  # Call the change method inside __init__

    def change(self):
        self.value = 'End'

# Create an instance of Modifier
item = Modifier()

# Print the value attribute
print(item.value)
```

```python
options = {1: 'Option A', 2: ['Option B', 'Option C']}  # Use dictionary with key-value pairs

# Insert 'Option B.1' into the second position of the list at options[2]
options[2].insert(1, 'Option B.1')  # ['Option B', 'Option B.1', 'Option C']
print(options[2], end=' ')  # Prints the modified list

# Convert options[1] to a list, append '!', and then join the list back into a string
choice_list = list(options[1])  # Convert string 'Option A' into a list of characters
choice_list.append('!')  # Append '!' to the list of characters
options[1] = ''.join(choice_list)  # Join the list back into a string
print(options[1])  # Print the modified string
```
